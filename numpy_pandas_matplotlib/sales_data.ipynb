```python
pd.pivot_table(data,index=["Name","Rep","Manager"],values=['Sales','Quantity'],aggfunc=np.sum)
```

```python
pd.pivot_table(data,index=["Product"],values=["Quantity","Sales"],aggfunc=np.sum)
```

```python
pd.pivot_table(data,index=["Product"],values=["Quantity","Sales"],margins='false',aggfunc=np.sum)
```

```python
pd.pivot_table(data,index=["Manager"],columns=["Product"],values=["Quantity","Sales"],aggfunc=np.sum)
```

```python
pd.pivot_table(data,index=["Manager"],columns=["Product"],values=["Quantity","Sales"],fill_value=0,aggfunc=np.sum)
```

```python
pd.pivot_table(data,index=["Manager"],columns=["Product"],values=["Quantity"],fill_value=0,aggfunc=[np.sum,np.mean,np.max])
```

```python
table = pd.pivot_table(data,index=["Manager"],columns=["Product"],values=["Quantity"],fill_value=0,aggfunc=[np.sum,np.mean,np.max])
table.query("Manager == 'Fred Anderson'")
```
