```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("https://raw.githubusercontent.com/d4dipdas/DANLC/main/student_data.csv")
```

```python
data
```

```python
table = data.groupby('Major')['City'].count()
table
```

```python
plt.pie(table,autopct='%1.1f%%')
plt.title("City wise student count")
plt.show()
```
