# 6. Check For Duplicate Data and Drop Them

```python
data = data.drop_duplicates()
```

# 7. Find Out Number of Courses Per Subjects

```python
num_courses = data['subject'].value_counts()
num_courses
```

```python
plt.bar(num_courses.index, num_courses.values,color='orange')
plt.xlabel('Subjects')
plt.ylabel('Counts')
plt.xticks(rotation = 65)
plt.title('Count Plot of Subjects')
plt.show()
```

# 8. For Which Levels, Udemy Courses Providing The Courses

```python
level_wise_count = data['level'].value_counts()
level_wise_count
```

```python
plt.barh(level_wise_count.index, level_wise_count.values, color='y')
plt.xlabel("Counts")
plt.ylabel("Level Names")
plt.title("Count according to levels")
plt.show()
```

# 9. Display The Count of Paid and Free Courses 

```python
paid_or_free = data['is_paid'].value_counts()
paid_or_free
```

```python
levels = ["Paid","Unpaid"]
plt.bar(levels, paid_or_free.values,color='black')
plt.xlabel("Is Paid")
plt.ylabel("Count")
plt.title("Count according to paid or not")
plt.show()
```

# 10. Which Course Has More Lectures (Free or Paid)?

```python
result = data.groupby(['is_paid'])['num_lectures'].mean()
print(result)
```

```python
plt.pie(result, autopct = '%1.1f%%')
plt.title("PAID vs FREE")
plt.show()
```

```python
print("Paid courses have more number of lectures")
```

# 11. Which Courses Have A Higher Number of Subscribers Free or Paid?

```python
grouped_data = data.groupby(['is_paid'])['num_subscribers'].sum()
grouped_data
```

```python
labels = ['UNPAID','PAID']
plt.bar(labels,grouped_data.values)
plt.xlabel("IS PAID")
plt.ylabel("Number of subscribers")
plt.title("Number of subscribers according to course paid or not")
plt.show()
```

# 12. Which Level Has The Highest Number of Subscribers?

```python
levelWise_data = data.groupby(['level'])['num_subscribers'].sum()
print(levelWise_data)

plt.barh(levelWise_data.index,levelWise_data.values,color='g')
plt.xlabel("Number of subscribers")
plt.ylabel("Levels")
plt.title("Level vs No. of subscribers")
plt.show()
```

```python
print("All Levels has highest number of subscribers")
```

# 13. Find Most Popular Course Title

```python
data[data['num_subscribers'].max() == data['num_subscribers']]['course_title']
```

# 14. Display 10 Most Popular Courses As Per Number of Subscribers

```python
top_10 = data.sort_values(by="num_subscribers",ascending=False).head(10)
top_10
```

```python
plt.barh(top_10['course_title'],top_10['num_subscribers'],color='y')
plt.xlabel("No. of subscribers")
plt.ylabel("Course title")
plt.title("TOP 10 Courses based on subscribers")
plt.show()
```

# 15. Find The Course Which Is Having The Highest Number of Reviews.

```python
data.sort_values(by='num_reviews').tail(1)
```

```python
sorted_reviews = data.sort_values(by='num_reviews').tail(5)
plt.figure(figsize=(11,5))
plt.bar(sorted_reviews['course_title'],sorted_reviews['num_reviews'],color='green')
plt.xlabel("Course Title")
plt.ylabel("Number of Reviews")
plt.title("Top 5 Courses based on Reviews")
plt.xticks(rotation=75)
plt.show()
```

# 16. Does Price Affect the Number of Reviews?

```python
plt.scatter(data['price'],data['num_reviews'])
plt.xlabel("Price")
plt.ylabel("Number of Reviews")
plt.title("Price vs No. of Reviews")
plt.show()

print("\n\nSo we can clearly see that price is directly proportional to number of reviews")
print("As the price is increasing, the number of reviews are also increasing")
```

# 17. Find Total Number of Courses Related To Python

```python
print("Courses related to Python\n")
python_courses = data[data['course_title'].str.contains("python",case=False)]
python_courses
```

```python
print("Total number of courses related to python are:",end="")
len(python_courses)
```

# 18. Display 10 Most Popular Python Courses As Per Number of Subscribers

```python
top_10_python = python_courses.sort_values(by='num_subscribers',ascending=False).head(10)
top_10_python
```

```python
plt.barh(top_10_python['course_title'],top_10_python['num_subscribers'],color='aqua')
plt.xlabel("Number of subscribers")
plt.ylabel("Course Title")
plt.title("Top 10 Python Courses based on subscribers")
plt.show()
```

# 19. In Which Year The Highest Number of Courses Were Posted?

```python
data['Year'] = data['published_timestamp'].dt.year
data.head(1)
```

```python
yearwise_count = data.groupby('Year')['course_title'].count()
yearwise_count
```

```python
plt.plot(yearwise_count.index,yearwise_count.values,color='red')
plt.xlabel("Year")
plt.ylabel("Number of courses")
plt.title("Number of courses vs Year")
plt.show()
```
